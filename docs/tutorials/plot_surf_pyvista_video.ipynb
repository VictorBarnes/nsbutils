```python
from importlib.resources import files
import matplotlib.pyplot as plt
import nibabel as nib
from neuromodes.io import fetch_surf
from neuromodes import EigenSolver
from nsbutils.utils import unmask
from nsbutils.plotting_pyvista import plot_surf, plot_surf_video
```

```python
species = 'human'
den = "32k"

surf_lh_data, medmask_lh = fetch_surf(density=den, hemi="L", species=species)
surf_lh = {"vertices": surf_lh_data.vertices, "faces": surf_lh_data.faces}
surfs = {"lh": surf_lh}
rois = {"lh": medmask_lh}
```

```python
solver_lh = EigenSolver(surf_lh, medmask_lh)
solver_lh.solve(n_modes=100)

dt = 1e-4
nt = 500
bold = solver_lh.simulate_waves(dt=dt, nt=nt)
bold_unmasked = unmask(bold, medmask_lh)
```

```python
file = plot_surf_video(
    surf=surf_lh, 
    rois=medmask_lh, 
    data_timeseries=bold_unmasked, 
    framerate=24,
    cmap="seismic",
    cbar=True
)
```
